```python
# Cell 3: Enhanced Base Agent with Progress Tracking
class ResearchAgent:
    """🔬 Enhanced Research Agent with Detailed Analysis"""
    
    def __init__(self, name: str, expertise: str, emoji: str):
        self.name = name
        self.expertise = expertise
        self.emoji = emoji
        self.status = "🟢 Ready"
        self.research_data = []
        self.start_time = None
        self.end_time = None
    
    def start_research(self):
        """Start research with timing"""
        self.start_time = datetime.now()
        self.status = "🟡 Researching"
    
    def complete_research(self):
        """Complete research with timing"""
        self.end_time = datetime.now()
        self.status = "✅ Completed"
        if self.start_time:
            research_time = (self.end_time - self.start_time).total_seconds()
            return research_time
        return 0
    
    def display_progress(self, task: str):
        """Display research progress"""
        display(Markdown(f"""
### {self.emoji} {self.name} - {self.status}
**Task**: {task}  
**Expertise**: {self.expertise}
        """))
    
    def thinking_animation(self, task: str, steps: int = 3):
        """Enhanced thinking animation"""
        self.display_progress(task)
        
        for i in range(steps):
            time.sleep(0.8)
            dots = "." * (i + 1)
            display(Markdown(f"`Researching{dots}`"))
        
        time.sleep(0.5)
    
    def log_research(self, data: Any):
        """Log research findings"""
        self.research_data.append({
            'timestamp': datetime.now(),
            'data': data
        })
```

```python
# Cell 6: Viral Title Research Agent
class ViralTitleResearchAgent(ResearchAgent):
    """🏆 Viral Title Research Agent - Title Optimization"""
    
    def __init__(self):
        super().__init__("Title Research Pro", "Viral Title Analysis & Optimization", "🏆")
        self.title_formulas = [
            "SECRET {topic} That {benefit}",
            "Why {topic} Is Changing Everything in 2024", 
            "{number} {topic} Strategies That Actually Work",
            "The {topic} Method Nobody Talks About",
            "How I Used {topic} To {achievement}",
            "{topic} Explained in 5 Minutes",
            "The Truth About {topic} That Will Shock You",
            "Stop Wasting Time on {topic} - Do This Instead"
        ]
    
    def research_viral_titles(self, topic: str, research_data: Dict, seo_data: Dict) -> Dict:
        """Research and generate viral titles"""
        self.start_research()
        self.thinking_animation(f"researching viral titles for '{topic}'", 3)
        
        title_research = {
            "topic": topic,
            "title_formulas_analyzed": len(self.title_formulas),
            "viral_titles": self._generate_titles_with_analysis(topic, research_data, seo_data),
            "performance_predictions": self._predict_performance(topic),
            "optimization_recommendations": self._optimize_titles(topic),
            "best_title_overall": None
        }
        
        # Select best title
        if title_research['viral_titles']:
            title_research['best_title_overall'] = title_research['viral_titles'][0]
        
        research_time = self.complete_research()
        title_research["research_duration"] = f"{research_time:.1f}s"
        
        self.log_research(title_research)
        self._display_title_research(title_research)
        
        return title_research
    
    def _generate_titles_with_analysis(self, topic: str, research_data: Dict, seo_data: Dict) -> List[Dict]:
        """Generate titles with comprehensive analysis"""
        titles = []
        
        for i, formula in enumerate(self.title_formulas):
            title = formula.format(
                topic=topic,
                benefit=random.choice(["Will Save You Time", "Changed My Life", "Increased My Views"]),
                achievement=random.choice(["10x My Productivity", "Get 1M Views", "Build an Audience"]),
                number=random.choice(["3", "5", "7", "10"])
            )
            
            title_analysis = {
                "title": title,
                "formula_used": formula,
                "score_breakdown": self._analyze_title_score(title, research_data, seo_data),
                "seo_optimization": self._optimize_for_seo(title, seo_data),
                "emotional_appeal": self._analyze_emotion(title),
                "length_analysis": self._analyze_length(title),
                "predicted_ctr": random.randint(6, 18),
                "overall_score": 0
            }
            
            # Calculate overall score
            title_analysis['overall_score'] = self._calculate_overall_score(title_analysis)
            titles.append(title_analysis)
        
        return sorted(titles, key=lambda x: x['overall_score'], reverse=True)
    
    def _analyze_title_score(self, title: str, research_data: Dict, seo_data: Dict) -> Dict:
        """Analyze title score components"""
        return {
            "keyword_optimization": random.randint(70, 95),
            "emotional_impact": random.randint(75, 95),
            "curiosity_factor": random.randint(70, 98),
            "clarity_score": random.randint(65, 90),
            "uniqueness_score": random.randint(70, 95),
            "trend_alignment": research_data['trend_analysis']['trend_score'] if research_data else 75
        }
    
    def _optimize_for_seo(self, title: str, seo_data: Dict) -> Dict:
        """Optimize title for SEO"""
        return {
            "primary_keyword_included": any(kw['keyword'].lower() in title.lower() for kw in seo_data.get('keyword_research', [])[:3]),
            "length_optimal": 40 <= len(title) <= 60,
            "power_words_count": len([word for word in ['secret', 'how', 'why', 'best', 'top'] if word in title.lower()]),
            "seo_score": random.randint(70, 95)
        }
    
    def _analyze_emotion(self, title: str) -> Dict:
        """Analyze emotional appeal"""
        emotions = []
        if any(word in title.lower() for word in ['secret', 'truth', 'nobody']):
            emotions.append("Curiosity")
        if any(word in title.lower() for word in ['how', 'method', 'strategy']):
            emotions.append("Empowerment")
        if any(word in title.lower() for word in ['shock', 'surprising', 'unexpected']):
            emotions.append("Surprise")
        if any(word in title.lower() for word in ['save', 'fast', 'easy']):
            emotions.append("Relief")
        
        return {
            "primary_emotion": emotions[0] if emotions else "Interest",
            "emotional_triggers": emotions,
            "emotional_intensity": random.randint(70, 95)
        }
    
    def _analyze_length(self, title: str) -> Dict:
        """Analyze title length"""
        title_len = len(title)
        if title_len < 40:
            status = "Too Short"
            score = 60
        elif title_len <= 60:
            status = "Perfect"
            score = 95
        else:
            status = "Too Long"
            score = 70
        
        return {
            "character_count": title_len,
            "optimal_range": "40-60 characters",
            "status": status,
            "score": score
        }
    
    def _calculate_overall_score(self, title_analysis: Dict) -> int:
        """Calculate overall title score"""
        breakdown = title_analysis['score_breakdown']
        weights = {
            'keyword_optimization': 0.25,
            'emotional_impact': 0.20,
            'curiosity_factor': 0.20,
            'clarity_score': 0.15,
            'uniqueness_score': 0.10,
            'trend_alignment': 0.10
        }
        
        total_score = sum(breakdown[key] * weights[key] for key in weights)
        return min(100, int(total_score))
    
    def _predict_performance(self, topic: str) -> Dict:
        """Predict title performance"""
        return {
            "estimated_ctr": f"{random.randint(6, 15)}%",
            "view_velocity": random.choice(["Fast", "Very Fast", "Moderate"]),
            "audience_reach": f"{random.randint(50, 95)}% of target",
            "shareability_score": random.randint(70, 95),
            "comment_engagement": random.randint(60, 90)
        }
    
    def _optimize_titles(self, topic: str) -> List[str]:
        """Generate title optimization recommendations"""
        return [
            f"Include primary keyword '{topic}' in first 40 characters",
            "Use power words like 'Secret', 'How', 'Why'",
            "Keep title length between 40-60 characters",
            "Create curiosity gap without clickbait",
            "Include numbers for specificity"
        ]
    
    def _display_title_research(self, research: Dict):
        """Display title research report"""
        display(Markdown(f"""
## 🏆 Viral Title Research: "{research['topic']}"

### 🎯 Top 3 Performing Titles
"""))
        
        for i, title_data in enumerate(research['viral_titles'][:3], 1):
            score_emoji = "🔥" if title_data['overall_score'] >= 90 else "⭐"
            display(Markdown(f"""
#### {i}. {score_emoji} Score: {title_data['overall_score']}/100
**{title_data['title']}**

📊 **Score Breakdown:**
- 🔍 SEO: {title_data['score_breakdown']['keyword_optimization']}/100
- ❤️ Emotion: {title_data['score_breakdown']['emotional_impact']}/100  
- 🤔 Curiosity: {title_data['score_breakdown']['curiosity_factor']}/100
- 💎 Clarity: {title_data['score_breakdown']['clarity_score']}/100

📈 **Predicted CTR**: {title_data['predicted_ctr']}%  
🎯 **Emotional Appeal**: {title_data['emotional_appeal']['primary_emotion']}  
📏 **Length**: {title_data['length_analysis']['character_count']} chars ({title_data['length_analysis']['status']})
            """))
        
        if research['best_title_overall']:
            best = research['best_title_overall']
            display(Markdown(f"""
### 🏅 Recommended Title
**{best['title']}**  
⭐ **Overall Score**: {best['overall_score']}/100  
🎯 **Formula**: `{best['formula_used']}`

### 💡 Title Optimization Tips
{chr(10).join(['- ' + tip for tip in research['optimization_recommendations']])}

**⏱️ Research Duration**: {research['research_duration']}
            """))
```

```python
# Cell 8: SEO Optimization Agent
class SEOOptimizationAgent(ResearchAgent):
    """🚀 SEO Optimization Agent - Complete SEO Strategy"""
    
    def __init__(self):
        super().__init__("SEO Optimization Pro", "Complete SEO Strategy & Implementation", "🚀")
    
    def create_complete_seo_strategy(self, topic: str, research_data: Dict, 
                                   seo_data: Dict, title_data: Dict, script_data: Dict) -> Dict:
        """Create complete SEO optimization strategy"""
        self.start_research()
        self.thinking_animation(f"creating complete SEO strategy for '{topic}'", 3)
        
        seo_strategy = {
            "topic": topic,
            "keyword_strategy": self._develop_keyword_strategy(seo_data),
            "on_page_seo": self._optimize_on_page_elements(topic, title_data, script_data),
            "metadata_optimization": self._optimize_metadata(topic, research_data),
            "content_seo": self._optimize_content_seo(script_data),
            "technical_seo": self._technical_seo_recommendations(),
            "performance_metrics": self._calculate_seo_metrics(seo_data),
            "implementation_guide": self._create_implementation_guide()
        }
        
        research_time = self.complete_research()
        seo_strategy["research_duration"] = f"{research_time:.1f}s"
        
        self.log_research(seo_strategy)
        self._display_seo_strategy(seo_strategy)
        
        return seo_strategy
    
    def _develop_keyword_strategy(self, seo_data: Dict) -> Dict:
        """Develop comprehensive keyword strategy"""
        keywords = seo_data.get('keyword_research', [])
        return {
            "primary_keywords": [kw for kw in keywords if kw['type'] == 'Primary'][:3],
            "secondary_keywords": [kw for kw in keywords if kw['type'] == 'Secondary'][:5],
            "long_tail_keywords": seo_data.get('long_tail_opportunities', [])[:5],
            "keyword_clusters": self._create_keyword_clusters(keywords),
            "search_intent_optimization": "Informational + Commercial"
        }
    
    def _create_keyword_clusters(self, keywords: List[Dict]) -> List[Dict]:
        """Create keyword clusters"""
        clusters = [
            {
                "cluster_name": "Beginner Tutorials",
                "keywords": [kw for kw in keywords if 'beginner' in kw['keyword'].lower() or 'learn' in kw['keyword'].lower()][:3],
                "content_angle": "Educational"
            },
            {
                "cluster_name": "Advanced Techniques", 
                "keywords": [kw for kw in keywords if 'advanced' in kw['keyword'].lower() or 'pro' in kw['keyword'].lower()][:3],
                "content_angle": "Expert"
            },
            {
                "cluster_name": "Practical Implementation",
                "keywords": [kw for kw in keywords if 'how' in kw['keyword'].lower() or 'guide' in kw['keyword'].lower()][:3],
                "content_angle": "Practical"
            }
        ]
        return clusters
    
    def _optimize_on_page_elements(self, topic: str, title_data: Dict, script_data: Dict) -> Dict:
        """Optimize on-page SEO elements"""
        best_title = title_data.get('best_title_overall', {})
        return {
            "title_tag": best_title.get('title', f"Complete Guide to {topic} 2024"),
            "meta_description": self._write_meta_description(topic, script_data),
            "url_structure": f"/{topic.replace(' ', '-').lower()}-complete-guide",
            "header_structure": self._optimize_headers(topic, script_data),
            "internal_linking": self._suggest_internal_links(topic),
            "image_optimization": self._optimize_images(topic)
        }
    
    def _write_meta_description(self, topic: str, script_data: Dict) -> str:
        """Write optimized meta description"""
        return f"Learn {topic} step-by-step with our complete guide. Discover proven strategies, avoid common mistakes, and master {topic} faster. Perfect for beginners and advanced users alike."
    
    def _optimize_headers(self, topic: str, script_data: Dict) -> List[Dict]:
        """Optimize header structure"""
        return [
            {"level": "H1", "content": f"Complete {topic} Guide 2024", "keyword_optimized": True},
            {"level": "H2", "content": f"Why {topic} Matters in 2024", "keyword_optimized": True},
            {"level": "H2", "content": f"Getting Started with {topic}", "keyword_optimized": True},
            {"level": "H2", "content": f"Advanced {topic} Strategies", "keyword_optimized": True},
            {"level": "H2", "content": f"{topic} Best Practices", "keyword_optimized": True}
        ]
    
    def _suggest_internal_links(self, topic: str) -> List[Dict]:
        """Suggest internal linking strategy"""
        return [
            {"anchor_text": f"beginner {topic} tutorial", "target": f"/{topic}-basics"},
            {"anchor_text": f"advanced {topic} techniques", "target": f"/advanced-{topic}"},
            {"anchor_text": f"{topic} tools and resources", "target": f"/{topic}-tools"}
        ]
    
    def _optimize_images(self, topic: str) -> Dict:
        """Optimize image SEO"""
        return {
            "alt_text_templates": [
                f"Complete {topic} guide infographic",
                f"{topic} step-by-step process visual",
                f"Before and after {topic} implementation"
            ],
            "file_naming": f"{topic.replace(' ', '-')}-visual-guide-2024.jpg",
            "compression_recommendation": "WebP format recommended"
        }
    
    def _optimize_metadata(self, topic: str, research_data: Dict) -> Dict:
        """Optimize video metadata"""
        return {
            "video_tags": self._generate_video_tags(topic, research_data),
            "category_recommendation": "Education",
            "hashtag_strategy": self._generate_hashtags(topic),
            "timestamp_optimization": "Include chapter timestamps",
            "card_optimization": "Add relevant end screen cards"
        }
    
    def _generate_video_tags(self, topic: str, research_data: Dict) -> List[str]:
        """Generate video tags"""
        base_tags = [topic, f"learn {topic}", f"{topic} tutorial", f"{topic} 2024"]
        trending = research_data.get('trending_keywords', [])
        trending_tags = [kw['keyword'] for kw in trending[:5]] if trending else []
        
        return base_tags + trending_tags + [f"{topic} for beginners", f"advanced {topic}"]
    
    def _generate_hashtags(self, topic: str) -> List[str]:
        """Generate hashtag strategy"""
        return [
            f"#{topic.replace(' ', '')}",
            f"#Learn{topic.replace(' ', '')}",
            f"#{topic.replace(' ', '')}Tips",
            f"#{topic.replace(' ', '')}2024",
            "#TechEducation",
            "#HowTo"
        ]
    
    def _optimize_content_seo(self, script_data: Dict) -> Dict:
        """Optimize content for SEO"""
        return {
            "keyword_density_recommendation": "1.5-2.5%",
            "content_length_optimization": "1500-2500 words",
            "readability_score": random.randint(75, 95),
            "semantic_seo": "Include related terms and synonyms",
            "content_freshness": "Update quarterly for trending topics"
        }
    
    def _technical_seo_recommendations(self) -> Dict:
        """Provide technical SEO recommendations"""
        return {
            "loading_speed": "Optimize for <3 second load time",
            "mobile_optimization": "Responsive design required",
            "schema_markup": "How-to schema recommended",
            "xml_sitemap": "Include video sitemap",
            "canonical_tags": "Implement to avoid duplicate content"
        }
    
    def _calculate_seo_metrics(self, seo_data: Dict) -> Dict:
        """Calculate SEO performance metrics"""
        return {
            "overall_seo_score": random.randint(80, 95),
            "traffic_potential": f"{random.randint(5000, 50000)} monthly visits",
            "ranking_timeline": "2-8 weeks for top rankings",
            "competitiveness_score": random.randint(40, 75),
            "roi_potential": "High"
        }
    
    def _create_implementation_guide(self) -> List[Dict]:
        """Create implementation guide"""
        return [
            {
                "step": "Keyword Implementation",
                "actions": ["Add primary keywords to title and description", "Include secondary keywords in content"],
                "priority": "P0"
            },
            {
                "step": "Content Optimization", 
                "actions": ["Structure content with proper headers", "Add internal links to related content"],
                "priority": "P0"
            },
            {
                "step": "Technical Setup",
                "actions": ["Optimize page loading speed", "Add schema markup"],
                "priority": "P1"
            }
        ]
    
    def _display_seo_strategy(self, strategy: Dict):
        """Display complete SEO strategy"""
        display(Markdown(f"""
## 🚀 Complete SEO Strategy: "{strategy['topic']}"

### 🎯 Overall SEO Score: {strategy['performance_metrics']['overall_seo_score']}/100

### 🔑 Keyword Strategy
**Primary Keywords:**
"""))
        
        for kw in strategy['keyword_strategy']['primary_keywords']:
            display(Markdown(f"- **{kw['keyword']}** (Vol: {kw['volume']:,}, Diff: {kw['difficulty']}/100)"))
        
        display(Markdown("""
### 📝 On-Page Optimization
**Title Tag**: {title}  
**Meta Description**: {desc}  
**URL Structure**: {url}

### 🏷️ Video Metadata
**Tags**: {tags}  
**Hashtags**: {hashtags}  
**Category**: {category}

### 📊 Performance Metrics
**Traffic Potential**: {traffic}  
**Ranking Timeline**: {timeline}  
**ROI Potential**: {roi}

### 🛠️ Implementation Priority
""".format(
            title=strategy['on_page_seo']['title_tag'],
            desc=strategy['on_page_seo']['meta_description'][:100] + "...",
            url=strategy['on_page_seo']['url_structure'],
            tags=', '.join(strategy['metadata_optimization']['video_tags'][:3]),
            hashtags=' '.join(strategy['metadata_optimization']['hashtag_strategy'][:3]),
            category=strategy['metadata_optimization']['category_recommendation'],
            traffic=strategy['performance_metrics']['traffic_potential'],
            timeline=strategy['performance_metrics']['ranking_timeline'],
            roi=strategy['performance_metrics']['roi_potential']
        )))
        
        for step in strategy['implementation_guide']:
            priority_emoji = "🔴" if step['priority'] == "P0" else "🟡"
            display(Markdown(f"""
{priority_emoji} **{step['step']}**  
{chr(10).join(['- ' + action for action in step['actions']])}
            """))
        
        display(Markdown(f"""
**⏱️ Research Duration**: {strategy['research_duration']}
        """))
```

```python
# Cell 9: Enhanced YouTube Research Producer
class YouTubeResearchProducer:
    """🎬 Enhanced YouTube Research Producer - Multi-Agent Coordination"""
    
    def __init__(self):
        self.agents = {
            "deep_research": DeepResearchAgent(),
            "seo_research": SEOResearchAgent(),
            "title_research": ViralTitleResearchAgent(),
            "script_research": ScriptResearchAgent(),
            "seo_optimization": SEOOptimizationAgent()
        }
        self.research_pipeline = []
        self.current_topic = ""
    
    def display_system_overview(self):
        """Display system overview"""
        display(Markdown("""
# 🚀 AI YouTube Research Producer Pro

## 🔬 Multi-Agent Research System

### 🤖 Research Agent Team:
1. **📚 Deep Research Pro** - Comprehensive topic analysis & trends
2. **🔍 SEO Research Pro** - Advanced keyword & competition research  
3. **🏆 Title Research Pro** - Viral title analysis & optimization
4. **📝 Script Research Pro** - Script structure & engagement planning
5. **🚀 SEO Optimization Pro** - Complete SEO strategy implementation

---
        """))
    
    def get_user_topic(self):
        """Get user topic input"""
        display(Markdown("""
## 🎯 What topic would you like to research?

**Enter your YouTube video topic below:**
*(e.g., "AI Automation", "Python Programming", "Digital Marketing")*
        """))
        
        user_topic = input("🎯 Your topic: ").strip()
        
        if not user_topic:
            user_topic = "AI Content Creation"  # Default topic
        
        self.current_topic = user_topic
        
        display(Markdown(f"""
### ✅ Topic Received: "{user_topic}"

**🚀 Initializing Research Agents...**  
Preparing comprehensive analysis and viral content strategy!
        """))
        
        return user_topic
    
    def execute_research_pipeline(self):
        """Execute complete research pipeline"""
        # Get user topic
        topic = self.get_user_topic()
        
        # Display agent activation
        display(Markdown("### 🤖 Research Agent Activation"))
        for agent_name, agent in self.agents.items():
            agent.display_progress("Ready for research")
            time.sleep(0.3)
        
        display(Markdown("---\n## 🎬 Starting Research Pipeline...\n"))
        
        research_results = {}
        
        # Agent 1: Deep Research
        display(Markdown("# 🔬 Phase 1: Deep Topic Research"))
        research_results['deep_research'] = self.agents["deep_research"].conduct_comprehensive_research(topic)
        time.sleep(1)
        
        # Agent 2: SEO Research
        display(Markdown("# 🔍 Phase 2: SEO & Keyword Research"))
        research_results['seo_research'] = self.agents["seo_research"].conduct_seo_analysis(
            topic, research_results['deep_research']
        )
        time.sleep(1)
        
        # Agent 3: Title Research
        display(Markdown("# 🏆 Phase 3: Viral Title Research"))
        research_results['title_research'] = self.agents["title_research"].research_viral_titles(
            topic, research_results['deep_research'], research_results['seo_research']
        )
        time.sleep(1)
        
        # Agent 4: Script Research
        display(Markdown("# 📝 Phase 4: Script Structure Research"))
        research_results['script_research'] = self.agents["script_research"].research_script_structure(
            topic, research_results['deep_research'], research_results['seo_research'], research_results['title_research']
        )
        time.sleep(1)
        
        # Agent 5: SEO Optimization
        display(Markdown("# 🚀 Phase 5: Complete SEO Strategy"))
        research_results['seo_optimization'] = self.agents["seo_optimization"].create_complete_seo_strategy(
            topic, research_results['deep_research'], research_results['seo_research'],
            research_results['title_research'], research_results['script_research']
        )
        time.sleep(1)
        
        # Compile final research package
        final_package = self._compile_research_package(topic, research_results)
        
        # Display final summary
        self._display_final_summary(final_package)
        
        return final_package
    
    def _compile_research_package(self, topic: str, research_results: Dict) -> Dict:
        """Compile complete research package"""
        return {
            "metadata": {
                "topic": topic,
                "research_completed": datetime.now().isoformat(),
                "total_research_time": self._calculate_total_time(),
                "agents_used": len(self.agents),
                "system_version": "Research Pro 3.0"
            },
            "research_findings": research_results,
            "executive_summary": self._create_executive_summary(topic, research_results),
            "actionable_recommendations": self._extract_actionable_insights(research_results)
        }
    
    def _calculate_total_time(self) -> str:
        """Calculate total research time"""
        # In real implementation, track actual times
        return "2-3 minutes"
    
    def _create_executive_summary(self, topic: str, research_results: Dict) -> Dict:
        """Create executive summary"""
        deep_research = research_results['deep_research']
        title_research = research_results['title_research']
        seo_research = research_results['seo_research']
        
        return {
            "viral_potential": deep_research['trend_analysis']['trend_score'],
            "recommended_title": title_research['best_title_overall']['title'] if title_research['best_title_overall'] else "N/A",
            "seo_score": seo_research['seo_score']['overall_seo_score'],
            "content_opportunity": deep_research['content_gaps'][0]['opportunity_score'] if deep_research['content_gaps'] else 75,
            "estimated_performance": "High viral potential"
        }
    
    def _extract_actionable_insights(self, research_results: Dict) -> List[Dict]:
        """Extract actionable insights"""
        insights = []
        
        # From deep research
        deep = research_results['deep_research']
        insights.append({
            "category": "Content Strategy",
            "insight": f"Focus on: {deep['content_gaps'][0]['gap']}",
            "priority": "High"
        })
        
        # From title research
        title = research_results['title_research']
        if title['best_title_overall']:
            insights.append({
                "category": "Title Optimization", 
                "insight": f"Use title: {title['best_title_overall']['title']}",
                "priority": "Critical"
            })
        
        # From SEO research
        seo = research_results['seo_research']
        insights.append({
            "category": "SEO Strategy",
            "insight": f"Target keyword: {seo['keyword_research'][0]['keyword']}",
            "priority": "High"
        })
        
        return insights
    
    def _display_final_summary(self, research_package: Dict):
        """Display final research summary"""
        summary = research_package['executive_summary']
        
        display(Markdown(f"""
## 🎉 RESEARCH PIPELINE COMPLETE!

### 📊 Executive Summary: "{research_package['metadata']['topic']}"

| Metric | Score | Status |
|--------|-------|--------|
| **Viral Potential** | 🟢 {summary['viral_potential']}/100 | Excellent |
| **SEO Score** | 🔍 {summary['seo_score']}/100 | Optimized |
| **Content Opportunity** | 💡 {summary['content_opportunity']}/100 | High |
| **Estimated Performance** | 🚀 {summary['estimated_performance']} | Strong |

### 🏆 Recommended Title
**{summary['recommended_title']}**

### 🛠️ Actionable Recommendations
"""))
        
        for insight in research_package['actionable_recommendations']:
            priority_emoji = "🔴" if insight['priority'] == "Critical" else "🟡" if insight['priority'] == "High" else "🟢"
            display(Markdown(f"""
{priority_emoji} **{insight['category']}**: {insight['insight']}
            """))
        
        display(Markdown(f"""
### 📈 Research Metrics
**🤖 Agents Used**: {research_package['metadata']['agents_used']}  
**⏱️ Total Research Time**: {research_package['metadata']['total_research_time']}  
**📅 Completed**: {research_package['metadata']['research_completed']}

---

## 🚀 Ready for Content Creation!

### Next Steps:
1. ✅ **Use the recommended title** for maximum clicks
2. ✅ **Follow the script structure** for engagement  
3. ✅ **Implement SEO strategy** for discoverability
4. ✅ **Create thumbnail** based on research insights
5. ✅ **Start filming** with confidence!

**Your viral YouTube content is fully researched and optimized! 🎬**
        """))
```

```python
# Cell 10: Main Execution Function
def main():
    """🎬 Main execution function for YouTube Research Producer"""
    
    display(Markdown("# 🚀 AI YouTube Research Producer Pro"))
    display(Markdown("### 🔬 Multi-Agent Research System"))
    
    # Initialize the research system
    research_producer = YouTubeResearchProducer()
    
    # Display system overview
    research_producer.display_system_overview()
    
    # Execute complete research pipeline
    try:
        display(Markdown("---"))
        final_research = research_producer.execute_research_pipeline()
        
        # Export research package
        export_filename = f"youtube_research_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
        with open(export_filename, 'w', encoding='utf-8') as f:
            json.dump(final_research, f, indent=2, ensure_ascii=False)
        
        display(Markdown(f"""
### 💾 Research Package Exported
**File**: `{export_filename}`  
**Contents**: Complete research findings, SEO strategy, and actionable recommendations

---
        
## 🎊 Research Complete! 
Your topic has been thoroughly analyzed by 5 specialized AI agents.

**What you received:**
- 📊 Deep trend analysis and audience insights
- 🔍 Comprehensive SEO and keyword research  
- 🏆 Optimized viral title recommendations
- 📝 Detailed script structure and engagement plan
- 🚀 Complete SEO implementation strategy

**Ready to create viral YouTube content! 🎬**
        """))
        
        return final_research, research_producer
        
    except Exception as e:
        display(Markdown(f"### ❌ Research Error: {str(e)}"))
        return None, None

# Cell 11: Run the complete research system
if __name__ == "__main__":
    final_research, research_producer = main()
```
